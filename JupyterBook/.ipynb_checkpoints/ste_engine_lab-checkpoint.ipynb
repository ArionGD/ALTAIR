## 3. Data Ingestion & Indicator Calculation Pipeline
Fetch 1-year historical daily bars for candidate stocks and compute:
- **50-DMA & 200-DMA**: Medium and long-term trend regimes.
- **Wilder's RSI(14)**: Momentum gauge.
- **Average True Range (ATR 14)**: Volatility gauge for dynamic stops/targets.
- **Up/Down Volume Ratio (20-day)**: Institutional buying/selling pressure proxy.

```python
def calculate_indicators(df: pd.DataFrame) -> pd.DataFrame:
    """Calculates technical indicators for swing trading."""
    df = df.copy()
    
    # Moving Averages
    df['MA_20'] = df['Close'].rolling(window=20).mean()
    df['MA_50'] = df['Close'].rolling(window=50).mean()
    df['MA_200'] = df['Close'].rolling(window=200).mean()
    
    # Wilder's RSI (14)
    delta = df['Close'].diff()
    gain = delta.clip(lower=0)
    loss = -delta.clip(upper=0)
    avg_gain = gain.ewm(alpha=1/14, min_periods=14, adjust=False).mean()
    avg_loss = loss.ewm(alpha=1/14, min_periods=14, adjust=False).mean()
    rs = avg_gain / avg_loss.replace(0, np.nan)
    df['RSI'] = 100 - (100 / (1 + rs))
    
    # Average True Range (ATR 14)
    high_low = df['High'] - df['Low']
    high_close = (df['High'] - df['Close'].shift()).abs()
    low_close = (df['Low'] - df['Close'].shift()).abs()
    true_range = pd.concat([high_low, high_close, low_close], axis=1).max(axis=1)
    df['ATR'] = true_range.rolling(window=14).mean()
    
    # Up/Down Volume Ratio (20-day)
    up_days = (df['Close'].diff() > 0).astype(int)
    down_days = (df['Close'].diff() < 0).astype(int)
    up_vol = (df['Volume'] * up_days).rolling(window=20).sum()
    down_vol = (df['Volume'] * down_days).rolling(window=20).sum()
    total_vol = up_vol + down_vol
    df['UD_Vol_Ratio'] = np.where(total_vol > 0, ((up_vol - down_vol) / total_vol) * 100, 0)
    
    # Technical Momentum Score (-100 to +100)
    # Trend Regime (40%), RSI Direction (35%), Volume Pressure (25%)
    trend_score = np.where((df['Close'] > df['MA_50']) & (df['MA_50'] > df['MA_200']), 40,
                  np.where((df['Close'] < df['MA_50']) & (df['MA_50'] < df['MA_200']), -40, 0))
    
    rsi_score = np.where(df['RSI'] >= 60, 35,
                np.where(df['RSI'] <= 40, -35,
                ((df['RSI'] - 50) / 10) * 35))
    
    vol_score = np.clip((df['UD_Vol_Ratio'] / 50) * 25, -25, 25)
    
    df['Technical_Score'] = np.clip(trend_score + rsi_score + vol_score, -100, 100)
    
    return df

# Download and enrich sample ticker
ticker = "ICICIBANK.NS"
raw_data = yf.download(ticker, period="1y", interval="1d", progress=False)
if isinstance(raw_data.columns, pd.MultiIndex):
    raw_data.columns = [col[0] for col in raw_data.columns]

df_stock = calculate_indicators(raw_data)
print(f"[+] Downloaded and processed {len(df_stock)} trading days for {ticker}")
df_stock[['Close', 'MA_50', 'MA_200', 'RSI', 'ATR', 'UD_Vol_Ratio', 'Technical_Score']].tail(5)
```

## 4. Dual-Gate Swing Strategy & Backtest Simulator
Simulates real swing trading entries and exits:
- **Entry Trigger**: Technical Score $\ge +20$ AND Price above 20-DMA (Bullish Crossover).
- **Dynamic Stop-Loss**: Entry Price $- (1.5 \times \text{ATR})$.
- **Dynamic Target**: Entry Price $+ (3.0 \times \text{ATR})$ ($1:2$ Risk-to-Reward Ratio).
- **Max Holding Period**: 15 trading days (Time stop).

```python
def backtest_swing_strategy(
    df: pd.DataFrame,
    tech_threshold: float = 20.0,
    sl_atr_mult: float = 1.5,
    tp_atr_mult: float = 3.0,
    max_holding_days: int = 15
) -> dict:
    """Simulates swing trades based on technical momentum score and dynamic ATR risk brackets."""
    trades = []
    in_position = False
    entry_idx = 0
    entry_price = 0.0
    stop_loss = 0.0
    take_profit = 0.0
    entry_date = None
    
    for i in range(200, len(df)):
        row = df.iloc[i]
        prev_row = df.iloc[i - 1]
        date = df.index[i]
        
        if not in_position:
            # Long Entry Condition: Technical Score crosses threshold and price is above 20 DMA
            if row['Technical_Score'] >= tech_threshold and prev_row['Technical_Score'] < tech_threshold:
                if row['Close'] > row['MA_20'] and not pd.isna(row['ATR']):
                    in_position = True
                    entry_idx = i
                    entry_date = date
                    entry_price = float(row['Close'])
                    stop_loss = round(entry_price - (sl_atr_mult * float(row['ATR'])), 2)
                    take_profit = round(entry_price + (tp_atr_mult * float(row['ATR'])), 2)
        else:
            # Check Exit Conditions
            holding_period = i - entry_idx
            curr_high = float(row['High'])
            curr_low = float(row['Low'])
            curr_close = float(row['Close'])
            
            # 1. Target Hit
            if curr_high >= take_profit:
                pnl_pct = ((take_profit - entry_price) / entry_price) * 100
                trades.append({
                    'entry_date': str(entry_date)[:10],
                    'exit_date': str(date)[:10],
                    'entry_price': entry_price,
                    'exit_price': take_profit,
                    'pnl_pct': round(pnl_pct, 2),
                    'result': 'WIN',
                    'reason': 'TARGET_HIT',
                    'holding_days': holding_period
                })
                in_position = False
            # 2. Stop Loss Hit
            elif curr_low <= stop_loss:
                pnl_pct = ((stop_loss - entry_price) / entry_price) * 100
                trades.append({
                    'entry_date': str(entry_date)[:10],
                    'exit_date': str(date)[:10],
                    'entry_price': entry_price,
                    'exit_price': stop_loss,
                    'pnl_pct': round(pnl_pct, 2),
                    'result': 'LOSS',
                    'reason': 'STOP_LOSS',
                    'holding_days': holding_period
                })
                in_position = False
            # 3. Time Stop
            elif holding_period >= max_holding_days:
                pnl_pct = ((curr_close - entry_price) / entry_price) * 100
                trades.append({
                    'entry_date': str(entry_date)[:10],
                    'exit_date': str(date)[:10],
                    'entry_price': entry_price,
                    'exit_price': curr_close,
                    'pnl_pct': round(pnl_pct, 2),
                    'result': 'WIN' if pnl_pct > 0 else 'LOSS',
                    'reason': 'TIME_EXIT',
                    'holding_days': holding_period
                })
                in_position = False
                
    # Calculate scorecard
    if not trades:
        return {'total_trades': 0, 'win_rate': 0.0, 'trades': []}
        
    wins = [t for t in trades if t['result'] == 'WIN']
    losses = [t for t in trades if t['result'] == 'LOSS']
    total_pnl = sum(t['pnl_pct'] for t in trades)
    win_rate = (len(wins) / len(trades)) * 100
    avg_win = np.mean([t['pnl_pct'] for t in wins]) if wins else 0.0
    avg_loss = np.mean([t['pnl_pct'] for t in losses]) if losses else 0.0
    profit_factor = abs(sum(t['pnl_pct'] for t in wins) / sum(t['pnl_pct'] for t in losses)) if losses and sum(t['pnl_pct'] for t in losses) != 0 else 999.0
    
    return {
        'total_trades': len(trades),
        'wins': len(wins),
        'losses': len(losses),
        'win_rate_pct': round(win_rate, 2),
        'total_pnl_pct': round(total_pnl, 2),
        'avg_win_pct': round(avg_win, 2),
        'avg_loss_pct': round(avg_loss, 2),
        'profit_factor': round(profit_factor, 2),
        'trades': trades
    }

# Run baseline backtest
results = backtest_swing_strategy(df_stock, tech_threshold=20.0, sl_atr_mult=1.5, tp_atr_mult=3.0)
print("==================================================")
print(f"STE Strategy Scorecard for {ticker}")
print("==================================================")
print(f"Total Trades:   {results['total_trades']}")
print(f"Win Rate:       {results['win_rate_pct']}%")
print(f"Total Return:   {results['total_pnl_pct']}%")
print(f"Profit Factor:  {results['profit_factor']}")
print(f"Avg Win / Loss: +{results['avg_win_pct']}% / {results['avg_loss_pct']}%")

pd.DataFrame(results['trades'])
```

## 6. Multi-Asset Parameter Calibration (Grid Search Tuning)
Optimize parameters across multiple core universe stocks (`RELIANCE.NS`, `ICICIBANK.NS`, `SBIN.NS`, `TCS.NS`, `INFY.NS`).

```python
universe = ["RELIANCE.NS", "ICICIBANK.NS", "SBIN.NS", "TCS.NS", "INFY.NS"]
print(f"[*] Running parameter optimization across universe: {universe}")

calibration_results = []

# Test parameter grid
for sl_mult in [1.2, 1.5, 2.0]:
    for tp_mult in [2.5, 3.0, 4.0]:
        total_wins = 0
        total_trades = 0
        total_pnl = 0.0
        
        for sym in universe:
            stock_raw = yf.download(sym, period="1y", interval="1d", progress=False)
            if isinstance(stock_raw.columns, pd.MultiIndex):
                stock_raw.columns = [col[0] for col in stock_raw.columns]
            
            stock_df = calculate_indicators(stock_raw)
            res = backtest_swing_strategy(stock_df, tech_threshold=20.0, sl_atr_mult=sl_mult, tp_atr_mult=tp_mult)
            total_wins += res['wins']
            total_trades += res['total_trades']
            total_pnl += res['total_pnl_pct']
            
        if total_trades > 0:
            calibration_results.append({
                'SL_ATR': sl_mult,
                'TP_ATR': tp_mult,
                'Risk_Reward_Ratio': f"1:{tp_mult/sl_mult:.1f}",
                'Total_Trades': total_trades,
                'Aggregate_Win_Rate': round((total_wins / total_trades) * 100, 2),
                'Cumulative_Return_Pct': round(total_pnl, 2)
            })

df_calibration = pd.DataFrame(calibration_results).sort_values(by='Aggregate_Win_Rate', ascending=False)
print("[+] Parameter Calibration Scorecard:")
df_calibration
```
